## Topic 2 · Methods
*Walkthrough: `02 - Methods`*

### D5. Add a method

1. Add a method `summary(self)` to the class below. It **returns** a string like `R-001: 87.0/100.0`,
   built from `self.label`, `self.value` and `self.limit`.
2. Create a record and print the result of calling `.summary()` on it.

**Expected output:** `R-001: 87.0/100.0`

````python
# D5
class Record:
    def __init__(self, label, value, limit):
        self.label = label
        self.value = value
        self.limit = limit

    def summary(self): # defining the method 'summary()'
        """Returns a summary of the object"""
        return(f"{self.label}: {self.value}/{self.limit}")

r1 = Record("R-001", 87.0, 100.0)
print(r1.summary())
````

### D6. `self` is whichever object made the call

1. Copy your class from D5 and add a method `status(self)` that returns `"OVER LIMIT"` (100% or more),
   `"WARNING"` (90% or more) or `"OK"`, worked out from `self.value` and `self.limit`.
2. Create three records and print each one's `.status()`:

| Record | Expected output |
|---|---|
| `Record("R-001", 87.0, 100.0)` | `OK` |
| `Record("R-002", 120.0, 100.0)` | `OVER LIMIT` |
| `Record("R-003", 95.0, 100.0)` | `WARNING` |

````python
# D6
class Record:
    def __init__(self, label, value, limit):
        self.label = label
        self.value = value
        self.limit = limit

    def summary(self):
        return(f"{self.label}: {self.value}/{self.limit}")

    def status(self): # defining the method 'status'
        """Calculates percentage and returns its status"""
        percent = (self.value/self.limit) * 100 # calculating and storing the percentage
        if percent >= 100: # checking whether the percentage is over limit
            return "OVER LIMIT"
        elif percent >= 90: # checking whether the percentage is greater than or equal to 90
            return "WARNING"
        else:
            return "OK"

r1 = Record("R-001", 87.0, 100.0)
r2 = Record("R-002", 120.0, 100.0)
r3 = Record("R-003", 95.0, 100.0)

print(r1.status())
print(r2.status())
print(r3.status())
````

### D7. One method calls another

1. Copy your class from D6 and add a method `percent(self)` that returns `(self.value / self.limit) * 100`.
2. Change `status(self)` so it uses `self.percent()` instead of working out the percentage itself.
3. Create `Record("R-001", 87.0, 100.0)` and print its `.percent()` and `.status()`.

**Expected output:** `87.0`, then `OK`

````python
# D7
class Record:
    def __init__(self, label, value, limit):
        self.label = label
        self.value = value
        self.limit = limit

    def summary(self):
        return(f"{self.label}: {self.value}/{self.limit}")

    def status(self):
            percent = self.percent() # calling the method 'percent()' and storing its return value 
            if percent >= 100:
                return "OVER LIMIT"
            elif percent >= 90:
                return "WARNING"
            else:
                return "OK"

    def percent(self): # defining the method 'percent()'
        """Calculates and returns percentagse"""
        return (self.value/self.limit) * 100

r1 = Record("R-001", 87.0, 100.0)
print(r1.percent(), r1.status()) # calling the methods percent() and status() and outputing their results
````

### D8. A method can change its own object

1. Define a class `Checker`. In `__init__(self)`, set `self.over_limit_count = 0`.
2. Add a method `check(self, value, limit)` that adds 1 to `self.over_limit_count` when `value` is
   greater than `limit`.
3. Create one `Checker`, call `.check()` three times with `(87, 100)`, `(120, 100)` and `(130, 100)`,
   then print `over_limit_count`.

**Expected output:** `2`

````python
# D8
class Checker: 
    def __init__(self): # defining the class 'Checker'
        self.over_limit_count = 0

    def check(self, value, limit): # defining the method 'check()'
        """Checks if value is over limit and keeps count of when it is"""
        if value > limit: # checking if value is greater than the limit
            self.over_limit_count += 1

c = Checker() # creating an onject 'c'
c.check(87, 100) # calling the method 'check()'
c.check(120, 100)
c.check(130, 100)
print(c.over_limit_count)
````

### D11. Load objects from a file

`records_sample.txt` (in this `LAB` folder) has one record per line, like `R-001,87,100`.

1. Write `load_records(path)` that reads the file and returns a list of `Record` objects
   (convert `value` and `limit` with `float()`).
2. Loop over the list and print each record's label and status.

**Expected output:** the same three lines as D9.

````python
# D11
def load_records(path): # defining the function 'load_records()'
    """Loads objects from a file into a list and returns the list"""
    records = [] # creating an empty list
    with open(path, "r") as f: 
        for line in f: # loop to go through each line in the file
            label, value, limit = line.split(",") # reading each line, extracting the required data and storing it in its respective variable
            records.append(Record(label, float(value), float(limit))) # creating and adding the record to the list
    return records
      
records = load_records("records_sample.txt") # calling 'load_records()' and storing its result
for r in records: # loop to go through list
    print(r.label, r.status())
````

### C1. Count the records over the limit

Add a method `is_over(self)` to `Record` that returns `True` or `False`. Load `records_sample.txt` with
`load_records` from D11, use `is_over()` to count how many records are over their limit, and print a
one-line summary.

**Expected output:** `1 of 3 records are OVER LIMIT`

````python
# C1
class Record:
    def __init__(self, label, value, limit):
        self.label = label
        self.value = value
        self.limit = limit

    def summary(self):
        return(f"{self.label}: {self.value}/{self.limit}")

    def status(self):
            percent = self.percent()
            if percent >= 100:
                return "OVER LIMIT"
            elif percent >= 90:
                return "WARNING"
            else:
                return "OK"

    def percent(self):
        return (self.value/self.limit) * 100

    def is_over(self):
         """Checks and returns whether a record's value is over limit"""
         if self.status() == "OVER LIMIT": # calling the medthod 'status()' to check the status of the record
            return True
         else: return False

records = load_records("records_sample.txt") # calling load_records() and storing the list it returns
count = 0

for r in records: # looping through the list
     if r.is_over() == True: # calling 'is_over()' on each record in the list
       count += 1
print(f"{count} of {len(records)} records are OVER LIMIT")
````

### C2. An account that locks

Write a class `Account` with a `username` and a count of failed logins that starts at 0.
It has two methods:

- `fail(self)`: adds 1 to the failed logins
- `locked(self)`: returns `True` if there have been 3 or more failed logins

**Test it:**

| Code | Expected output |
|---|---|
| an account for `"anna"` with `.fail()` called 3 times, then `print(anna.locked())` | `True` |
| an account for `"ben"` with `.fail()` called once, then `print(ben.locked())` | `False` |

````python
# C2
class Account:
    def __init__(self, username): # defining the class 'Account'
        self.username = username
        self.count = 0

    def fail(self): 
        self.count += 1

    def locked(self):
        if self.count >= 3: # checking if there have been 3 or more failed logins
            return True
        else: return False
````

````python
anna = Account("anna") # creating an object 'anna'
anna.fail()
anna.fail()
anna.fail()
print(anna.locked())
````

````python
ben = Account("ben") # creating an object 'ben'
ben.fail()
print(ben.locked())
````
